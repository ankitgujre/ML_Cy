```python
# Preprocessing
# pip install scikit-learn
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.impute import SimpleImputer
```

```python
# create simple dataset
data = {
    "id": [1, 2, 3, 4, 5, 6],
    "age": [25, np.nan, 30, 35, np.nan, 100],
    "Salary": [50000, 60000, np.nan, 80000, 90000, np.nan],
    "city": ["NewYork", "London", np.nan, "Paris", "London", np.nan]
}
df = pd.DataFrame(data)
print("Original Data: ")
df
```

```python
# mean Impute
mean_imputer = SimpleImputer(strategy = 'mean')
df_mean = df.copy()
df_mean[['age','Salary']] = mean_imputer.fit_transform(df_mean[['age','Salary']])
print("Mean Imputer")
display(df_mean)
```

```python
# median imputer
median_imputer = SimpleImputer(strategy = 'median')
df_median = df.copy()
df_median[['age','Salary']] = median_imputer.fit_transform(df_median[['age','Salary']])
print("median Imputer")
display(df_median)
```

```python
# mode imputer
mode_imputer = SimpleImputer(strategy = "most_frequent")
df_mode = df.copy()
df_mode[['city']] = mode_imputer.fit_transform(df_mode[['city']])
print("mode imputer")
df_mode
```
